```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
```

```python
data = pd.read_csv("creditcard.csv")

print("Dataset Shape:", data.shape)
data.head()
```

```python
print(data.info())

print("\nMissing Values:")
print(data.isnull().sum())

print("\nDuplicate Rows:")
print(data.duplicated().sum())
```

```python
data = data.drop_duplicates()

print("Dataset Shape after removing duplicates:", data.shape)
```

```python
print(data["Class"].value_counts())
```

```python
class_counts = data["Class"].value_counts().sort_index()

percentages = class_counts / len(data) * 100

plt.figure(figsize=(8, 5))

bars = plt.bar(
    ["Normal", "Fraud"],
    percentages.values,
    color=["lightblue", "red"]
)

plt.xlabel("Class")
plt.ylabel("Percentage of Transactions")
plt.title("Percentage of Normal vs Fraudulent Transactions")

for bar, percentage in zip(bars, percentages.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{percentage:.3f}%",
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
X = data.drop("Class", axis=1)
y = data["Class"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training data shape:", X_train_scaled.shape)
print("Testing data shape :", X_test_scaled.shape)
```

```python
X_train_scaled = X_train_scaled.astype("float32")
X_test_scaled = X_test_scaled.astype("float32")
```

```python
X_train_normal = X_train_scaled[
    y_train.values == 0
]

print(
    "Normal transactions used for training:",
    X_train_normal.shape
)
```

```python
input_features = X_train_normal.shape[1]

autoencoder = Sequential([

    # Input layer
    Input(shape=(input_features,)),

    # Encoder
    Dense(64, activation="relu"),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(8, activation="relu"),

    # Decoder
    Dense(16, activation="relu"),
    Dense(32, activation="relu"),
    Dense(64, activation="relu"),

    # Output layer
    Dense(input_features, activation="linear")
])

autoencoder.summary()
```

```python
autoencoder.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

print("Autoencoder compiled successfully.")
```

```python
history = autoencoder.fit(
    X_train_normal,
    X_train_normal,

    epochs=5,

    batch_size=20000,

    validation_split=0.20,

    shuffle=True,

    verbose=1
)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(history.history["loss"],label="Training Loss")
plt.plot(history.history["val_loss"],label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Credit Card Autoencoder Training Loss")
plt.legend()

plt.show()
```

```python
X_train_reconstructed = autoencoder.predict(X_train_normal,batch_size=256,verbose=0)
print("Original shape:",X_train_normal.shape)
print("Reconstructed shape:",X_train_reconstructed.shape)
```

```python
normal_error = np.mean(np.square(X_train_normal - X_train_reconstructed),axis=1)

print("Normal reconstruction error:")
print(normal_error[:10])
```

```python
threshold = np.percentile(
    normal_error,
    95
)

print(
    "Anomaly Threshold:",
    threshold
)
```

```python
X_test_reconstructed = autoencoder.predict(
    X_test_scaled,
    batch_size=256,
    verbose=0
)

print(
    "Test reconstructed shape:",
    X_test_reconstructed.shape
)
```

```python
test_error = np.mean(
    np.square(
        X_test_scaled -
        X_test_reconstructed
    ),
    axis=1
)

print("First 10 test errors:")
print(test_error[:10])
```

```python
y_pred = (
    test_error > threshold
).astype(int)

print("Predicted classes:")
print(y_pred[:20])
```

```python
results = pd.DataFrame({
    "Actual_Class": y_test.values,
    "Reconstruction_Error": test_error,
    "Predicted_Class": y_pred
})

results.head(20)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)
    
print("Accuracy:", accuracy)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(
    cm,
    cmap="Blues"
)

plt.colorbar()

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.title("Confusion Matrix")

plt.xticks(
    [0, 1],
    ["Normal", "Fraud"]
)

plt.yticks(
    [0, 1],
    ["Normal", "Fraud"]
)

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.show()
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Normal",
            "Fraud"
        ]
    )
)
```

```python
fraud_transactions = results[
    results["Predicted_Class"] == 1
]

print(
    "Number of detected anomalies:",
    len(fraud_transactions)
)

fraud_transactions.head(20)
```
